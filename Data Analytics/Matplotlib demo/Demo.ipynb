```python
import seaborn as s
import matplotlib.pyplot as p
```

```python
month=['Jan','Feb','March','April']
sales=[100,150,120,200]
s.lineplot(x=month,y=sales)
p.show()
```

```python
month=['Jan','Feb','March','April']
sales=[100,150,120,200]
s.barplot(x=month,y=sales)
p.show()
```

```python
month=['Jan','Feb','March','April']
sales=[100,150,120,200]
s.scatterplot(x=month,y=sales)
p.show()
```

```python
month=['Jan','Feb','March','April']
sales=[100,150,120,200]
s.histplot(sales)
p.show()
```

### Countplot

```python
courses=['C','CPP','Java','Java','C','Python','C','Python','Java']
s.countplot(x=courses)
p.show()
```

### Heatmap
    values in the form of colors

```python
import pandas as pd
data={'Python':[80,90,70,60],
      'Java':[20,30,50,40],
      'SQL':[80,30,90,40]}
df=pd.DataFrame(data)
df
```

```python
c=df.corr()
s.heatmap(c)
p.show()
```
