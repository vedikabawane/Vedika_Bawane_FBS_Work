```python
import pandas as pd
```

```python
df1=pd.DataFrame({'ID':[101,102,103,104],
                 'NAME':['Vedika','Amisha','Ashawini','Vandana'],
                 'AGE':[22,21,19,40]})
df1
```

```python
df2=pd.DataFrame({'ID':[105,106,107],
                 'NAME':['Kartik','Rohan','Rahul'],
                 'AGE':[22,21,19]})
df2
```

```python
pd.concat([df1,df2])    #row ko combain kiya
```

```python
df3=pd.DataFrame({'ID':[106,107,108],
                 'NAME':['Kartik','Rohan','Rahul'],
                 'SALARY':[100,200,500]})
df3
```

```python
pd.concat([df1,df3])
```

```python
df4=pd.DataFrame({'ADDRESS':['Pune','Mumbai','Warora','Nagpur'],
                 'NAME':['DEV','DA','AI/ML','DS'],
                 'SALARY':[100,200,500,400]})
df4
```

```python
pd.concat([df1,df4],axis=1)
```

```python
emp=pd.DataFrame({'ID':[101,102,103,104],
                 'NAME':['Vedika','Amisha','Ashawini','Vandana'],
                 'AGE':[22,21,19,40],
                 'DID':[1,3,2,5]})
emp
```

```python
dept=pd.DataFrame({'DID':[1,2,3,4],
                 'DNAME':['TI','Admin','HR','Sales']})
dept
```

```python
  pd.merge(emp,dept)     #match data deta hai
```

```python
dept2=pd.DataFrame({'ID':[1,2,3,4],
                 'DNAME':['TI','Admin','HR','Sales']})
dept2
```

```python
 pd.merge(emp,dept2,left_on='DID',right_on='ID') 
```

```python
pd.merge(emp,dept,how='left')
```

```python
pd.merge(emp,dept,how='right')
```

```python
pd.merge(emp,dept,how='outer')
```
