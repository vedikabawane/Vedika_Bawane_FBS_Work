#### Lineplot
    1.Time series analysis
    2.Trends over time
    3.Upward/Downward trends
    4.IPL team scoresheet

```python
overs=np.arange(1,21)
team1_runs=np.random.randint(0,37,20)
team2_runs=np.random.randint(0,37,20)
```

```python
team1_runs     #add hote jana chahiye  -->  comminative function
```

```python
team1_score=np.cumsum(team1_runs)
team2_score=np.cumsum(team2_runs)
```

```python
team1_runs 
```

```python
team1_score
```

```python
plt.plot(overs,team1_score,color='red')
plt.show()
```

```python
plt.style.use('dark_background')
plt.plot(overs,team1_score,color='red',label='GT')
plt.plot(overs,team2_score,color='blue',label='MI')
plt.xlabel('Overs')
plt.ylabel('Runs')
plt.title('Scoresheet')
plt.legend()  
plt.show()
```
